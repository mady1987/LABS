```python
# test_configure_orian_workflow.py
from orian_mcp_server import configure_orian_workflow, OrianWorkflow, DataType, TaskType, AssignmentType

def test_configure():
    payload = OrianWorkflow(
        workflowName="SampleWorkflow2",
        description="Sample workflow for demonstration",
        app="Agent Orc",
        contexts=[{"key":"REQUESTER","description":"Person who requested","dataType":DataType.STRING}],
        tasks=[
            {
                "name":"SUBMIT_REQUEST","description":"Submit initial request",
                "taskType":TaskType.MANUAL,"assignmentType":AssignmentType.USER,"sequence":1,
                "outcomes":[{"name":"SUBMITTED","description":"Request submitted","nextTask":"REVIEW_REQUEST"}]
            },
            {
                "name":"REVIEW_REQUEST","description":"Review request",
                "taskType":TaskType.MANUAL,"assignmentType":AssignmentType.GROUP,"sequence":2,
                "outcomes":[{"name":"APPROVED","nextTask":None},{"name":"NEEDS_INFO","nextTask":"SUBMIT_REQUEST"}]
            }
        ],
        startInstance=False,
        initialContext={"REQUESTER":"alice@example.com"}
    )
    # set ORIAN_BASE_URL env before running this test
    res = configure_orian_workflow(payload)
    print(res)
    assert "ok" in res
```
