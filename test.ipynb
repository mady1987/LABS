```python
import os
from openai import OpenAI

# Fetch API key from environment variable
client = OpenAI() 

response = client.chat.completions.create(
    model="gpt-4o-mini",  # cheaper + fast
    messages=[
        {"role": "system", "content": "You have to execute the orchestration system to configure and manage workflows."},
        {"role": "user", "content": "Configure a workflow to automate data backup every night at 2 AM."}
    ]
)

print(response.choices[0].message.content)
```
